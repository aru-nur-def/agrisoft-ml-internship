## 7. Visualize All Training Metrics

This graph shows how the available training metrics change over epochs.
It can be used as a screenshot in the report section **7.5 Visualization of Results**.

```python
df.plot(figsize=(12, 6))
plt.title("Training Results")
plt.xlabel("Epoch")
plt.ylabel("Value")
plt.grid(True)
plt.show()
```

## 8. Plot Important Metrics Separately

This section tries to automatically find common YOLO training metrics and visualize them separately.
If some columns are not found, the code will skip them.

```python
possible_metrics = [
    "metrics/precision(B)",
    "metrics/recall(B)",
    "metrics/mAP50(B)",
    "metrics/mAP50-95(B)",
    "train/box_loss",
    "train/cls_loss",
    "val/box_loss",
    "val/cls_loss"
]

available_metrics = [col for col in possible_metrics if col in df.columns]

print("Available metrics found:")
print(available_metrics)

if available_metrics:
    df[available_metrics].plot(figsize=(12, 6))
    plt.title("Selected Training Metrics")
    plt.xlabel("Epoch")
    plt.ylabel("Value")
    plt.grid(True)
    plt.show()
else:
    print("No common YOLO metric columns found. Use the previous graph instead.")
```

## 9. Save Graph for Report

This cell saves the main training graph as an image file.
The file can be uploaded to GitHub or inserted into the report.

```python
plt.figure(figsize=(12, 6))
df.plot(figsize=(12, 6))
plt.title("Training Results")
plt.xlabel("Epoch")
plt.ylabel("Value")
plt.grid(True)
plt.savefig("training_results.png", dpi=300, bbox_inches="tight")
plt.show()

print("Saved file: training_results.png")
```
